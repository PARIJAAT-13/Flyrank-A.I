# ML-12 — 5-Minute Demo Outline

> 5 minutes total. Talk track below. All numbers are taken directly from

> paper. Nothing new is fabricated.

## 1. Question — what problem was being solved?  *(~45 sec)*

Content teams have tens of thousands of existing pages. Not every page is worth
refreshing — and editors cannot manually triage them all. The question this
project answers is: **which existing pages are most worth a human reviewer
spending time on, ranked, using only signals that are safe to publish**?

## 2. Method — what data, features, baseline, validation?  *(~75 sec)*

- **Data:** 30,000 anonymized content rows from the bundled starter dataset.
  Declining-label rate: 0.542. No titles, URLs, client names, domains, or
  keywords were used.
- **Features:** engagement-derived signals only — `days_with_impressions`,
  `log_impressions_90d`, `avg_position`, `content_age_days`, `char_count`,
  `word_count`, `log_clicks_90d`, `ctr`, `scroll_rate`, `days_with_sessions`.
- **Models compared:** decision_tree, logistic_regression, **random_forest**,
  and a transparent **baseline_rules** scorer. Best model was selected on
  `precision_at_50`.
- **Validation:** `client_holdout` split — a client's entire rows are held out
  so the score measures **intra-client generalization**, not row-level leakage.
  This matters because a model that "knows" a client will overstate quality.

## 3. One Chart — the strongest existing chart  *(~60 sec)*

**Chart: `outputs/charts/top_feature_importance.svg`** — *"Top model features"*.

This is the model's feature-importance bar chart from the Random Forest run.
What it shows:

- This is the chart to keep on screen because it answers "is this safe?"

## 4. One Honest Result — strongest evidence-backed result + limitation  *(~60 sec)*

**Result:** Random Forest reaches **Precision@50 = 0.740** vs the
`baseline_rules` scorer at **Precision@50 = 0.240** on the client-holdout
split — roughly a **3.1× lift** in how often the top-50 ranked rows actually
contain a declining page. Output queue: 30,000 rows ranked, with **3,602
high-confidence** rows ready for editorial review.

**Honest limitation:** This is a 30,000-row teaching slice. The metric is
directional and decision-support only — the project does not claim a deployed
causal effect on traffic. A real production version would need a longer
time window and a real editorial-outcome label.

## 5. One Recommendation — highest-priority actionable recommendation  *(~45 sec)*

**Wire the high-confidence band (3,602 rows) into editorial review as the
first production use.** That is the smallest, safest step that actually
exercises the model end-to-end: a human reviewer looks at the top of the
queue, checks the page, and either refreshes or passes. Every other action
(`refresh`, `refresh_and_review_ctr`, `refresh_and_review_engagement`,
`expand_and_refresh`) inherits its quality from how well that first loop
runs. Don't auto-publish from the queue — use it as a reviewer aid.
